```python
import os
import getpass
import subprocess
from pathlib import Path

REPO_URL = 'https://github.com/franektrafas/betclic.v1.git'
REPO_DIR = Path('/content/betclic.v1')
REPO_IS_PRIVATE = True
GITHUB_TOKEN = ''

clone_url = REPO_URL
if REPO_IS_PRIVATE and not GITHUB_TOKEN:
    GITHUB_TOKEN = getpass.getpass('GitHub token with read access to franektrafas/betclic.v1: ')

if REPO_IS_PRIVATE and GITHUB_TOKEN:
    clone_url = REPO_URL.replace('https://', f'https://{GITHUB_TOKEN}@')

if REPO_DIR.exists():
    subprocess.run(['git', '-C', str(REPO_DIR), 'pull'], check=True)
else:
    subprocess.run(['git', 'clone', clone_url, str(REPO_DIR)], check=True)

os.chdir(REPO_DIR)
print(f'Working directory: {REPO_DIR}')
```
